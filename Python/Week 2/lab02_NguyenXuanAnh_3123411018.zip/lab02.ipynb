```python
#bài 8: Autograd cơ bản
import torch
x = torch.tensor(3.6, requires_grad=True)
y = x* x
y.backward()

print("x = ", x)
print("y = ", y)
print("x.grad = ", x.grad)
print("Đạo hàm thủ công: ",3.6*2)
x.grad.zero_()
y = 3*x**2 + 2*x + 1
y.backward()
print("x.grad mới = ", x.grad)

print("Đổi hàm thành: ",y)
```

```python
#bài 9: Gradient descent tìm hệ số đa thức
import numpy as np
import torch
# tạo đa thức
polynomial = np.poly1d([2, -3, 5])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
#tạo ma trận đặt trưng cho đa thức bậc 2
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()# Xóa gradient cũ
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()# Tính gradient mới, lưu vào w.grad
    optimizer.step()# Dùng gradient để cập nhật w
print("Learned coefficients:")
print(w)
```

```python
#bài 10: Giải hệ phương trình bằng autograd
import random
import torch
A = torch.tensor(random.random(), requires_grad=True) # tạo ra một số ngẫu nhiên và cho phép tính gradient sau này
B = torch.tensor(random.random(), requires_grad=True)   
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000 # tạo ra 2000 lần lặp để tính toán giá trị tối ưu
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01) # tạo ra một đối tượng optimizer NAdam để sửa A, B, C, D với learning rate = 0.01
for _ in range(EPOCHS): # lập lại quá trình 2000 lần
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4 # tính tổng biến động của y1, y2, y3, y4 và lấy bình phương của nó để tính sqerr
    optimizer.zero_grad() # reset gradient về 0 để gradient cũ không bị cộng vào
    sqerr.backward() # tính gradient của sqerr đối với A, B, C, D
    optimizer.step() # cập nhật giá trị của A, B, C, D dựa trên gradient vừa tính được
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```
